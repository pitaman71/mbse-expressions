# 8.07 · C Types

C writes a type where an expression needs one: in a cast, `(uint8_t)200`, in `sizeof(int)`, in a compound literal,
`(int){1}`. What it writes there is a **type name** (ISO/IEC 9899, 6.7.7, "Type names"): a list of specifiers and
qualifiers, `const char`, then an abstract declarator, `*`, that says how the type is built from them. Tutorial 8
showed that a type name is a term of a dialect of its own, `C Types` (08.1); this page builds its terms from the
smallest up, the way tutorials 3 to 7 built expressions: a keyword, then specifiers and declarators, then structures
and enumerations, each printed back as C writes it; then what validation says of them, how C reads one in a cast, and
the editions they come from.

Like the rest of Part I, it starts from `dialects()`, a store with every dialect installed and defined. `S` is its holder
namespace, `store.Expressions`, and `amd64` x86-64's System V ABI, which a C family's holder names (tutorial 2).
`variable` writes a name in C's terms (tutorial 4), and `keyword` a type specifier that is a keyword, such as `int`,
each a spec the builders call, as tutorial 8's `uint8` is.

````typescript
import assert from "node:assert/strict";
import * as C from "@mbse/expressions/C/Expressions";
import * as CTypes from "@mbse/expressions/C/Types";
import { Standards, Symbolics, Terms } from "@mbse/expressions/Framework";
import { Repr } from "@mbse/schemas/Framework";
import { dialects } from "./toolkit.js";

const store = dialects();
const S = store.Expressions;
const amd64 = store.Instances.Expressions.SystemV.AMD64;
const variable = (name: string) => (d: any) => d.variable((v: any) => v.name(name));
const keyword = (word: string) => (s: any) => s.primitivetypespecifier((p: any) => p.keyword(word));
const DIALECT = CTypes.Expressions.DIALECT;

/** What `evaluate` says of `expression`: its value, or the reason it refuses it. */
function refused(evaluate: (expression: any) => unknown, expression: any): string {
  try {
    return String(evaluate(expression));
  } catch (error) {
    return (error as Error).message;
  }
}
````

<a id="T08.07.1"></a>
## 08.07.1 · A type name, `int`

```text
int
```

The smallest type name is one keyword. `C Types`' kind for a type name is `typeid`, as mbse-programs and C++'s grammar
call it: its `specifiers`, a list, and its `declarator`, absent here. A specifier is not a term but a **helper** (tutorial
6), a value its term holds: `int` is a `primitivetypespecifier` whose `keyword` is `int`. The dialect's holder,
`S.C.Types.Expression`, holds the type name as its root and names its ABI, as C's holder does, since what `int` is
depends on it (02.3). Its `Text.ToText`, `CTypes.Text.ToText`, prints it as C writes it:

````typescript
const int_ = store.transaction(() => S.C.Types.Expression((e: any) => e.root(
  (u: any) => u.typeid((i: any) => i.specifiers([keyword("int")]))).abi([amd64])));
const term = Terms.term(int_);
const form = Terms.form(term);
console.log(CTypes.Text.ToText(int_), "|", Terms.kindOf(term)[0], "of", Terms.dialectOf(term));
console.log(form.kind, "| specifiers:", ...(form.attributes.get("specifiers") as Terms.Part[]).map((p) => Repr.repr(p)),
  "| its terms:", form.arguments.length);
console.log(DIALECT.name(), DIALECT.tag, "| its kinds:", [...DIALECT.kinds().keys()].join(", "));
assert.ok(CTypes.Text.ToText(int_) === "int" && [...int_.abi].map((entry: any) => entry.abi)[0] === amd64);
````

Its kinds are two, `typeid` and `enumerator` (08.07.3), with `error` and `missing`, which every complete dialect has for
text it could not read (08.07.4). Everything else in a type name is a helper of `typeid`'s, twenty-two of them, each a part
of C's grammar for type names:

````typescript
console.log(CTypes.Expressions.HELPERS.join(", "));
````

<a id="T08.07.2"></a>
## 08.07.2 · Specifiers, then declarators

```text
unsigned long        const char*        char[4]        int(*)[4]        int(*)(int, ...)
```

Specifiers come first, in order, each a helper: two keywords, `unsigned long`; a qualifier, `const`, and a keyword. A
**declarator** follows them, a helper too, and nests as C's declarators do, from the outside in: `const char*` is a
`pointerdeclarator`; `char[4]` an `arraydeclarator`, whose `size` is C's term, the literal `4`; `int(*)[4]`, a pointer to
an array of four `int`s, is an `arraydeclarator` whose own declarator is the pointer, which the printer parenthesizes;
and `int(*)(int, ...)`, a pointer to a function, a `functiondeclarator` whose parameters are helpers again. A form reads
a type name as it reads any term (tutorial 6): its helpers among its attributes, and the terms they hold, such as an
array's size, its arguments, each a term of the dialect that reads it:

````typescript
const pointer = (d: any) => d.pointerdeclarator((p: any) => p);
const four = (t: any) => t.literal((r: any) => r.value((v: any) => v.as_scalar(4n)));
const names: any[] = store.transaction(() => [
  (i: any) => i.specifiers([keyword("unsigned"), keyword("long")]),
  (i: any) => i.specifiers([(s: any) => s.cvqualifier((q: any) => q.keyword("const")), keyword("char")]).declarator(pointer),
  (i: any) => i.specifiers([keyword("char")]).declarator((d: any) => d.arraydeclarator((a: any) => a.size(four))),
  (i: any) => i.specifiers([keyword("int")]).declarator((d: any) => d.arraydeclarator((a: any) => a.declarator(pointer).size(four))),
  (i: any) => i.specifiers([keyword("int")]).declarator((d: any) => d.functiondeclarator((f: any) => f
    .declarator((p: any) => p.parenthesizeddeclarator((q: any) => q.declarator(pointer)))
    .parameters([(p: any) => p.parameterdeclaration((q: any) => q.specifiers([keyword("int")])),
      (p: any) => p.ellipsisparameter((e: any) => e)]))),
].map((spec) => S.C.Types.Expression((e: any) => e.root((u: any) => u.typeid(spec)).abi([amd64]))));
for (const name of names) {
  const form = Terms.form(Terms.term(name));
  const declarator = form.attributes.get("declarator") as Terms.Part | undefined;
  console.log(CTypes.Text.ToText(name).padEnd(17), "specifiers:",
    (form.attributes.get("specifiers") as Terms.Part[]).map((s) => s.kind).join(", "),
    "| declarator:", declarator ? declarator.kind : "-",
    "| its terms:", form.arguments.map((a: any) => `${Terms.kindOf(a)[0]} of ${Terms.dialectOf(a)}`).join(", ") || "none");
}
assert.deepEqual(names.map((name) => CTypes.Text.ToText(name)), [
  "unsigned long", "const char*", "char[4]", "int(*)[4]", "int(*)(int, ...)"]);
````

<a id="T08.07.3"></a>
## 08.07.3 · Structures and enumerations

```text
struct S {}        enum Color {
                       RED = 1,
                       GREEN,
                       BLUE
                   }
```

A structure's or an enumeration's specifier may declare it, with a body: `struct S {}` is a `classspecifier` (its `key`
`struct`, or `union`) whose `body` is set. A structure's members are declarations, mbse-programs' statements, which a
type name does not hold. An enumeration's are its **enumerators**, each a term of `C Types`' own, its kind `enumerator`:
a name, and a `value`, C's term, where one is written. So `enum Color { RED = 1, GREEN, BLUE }` is a `typeid` whose
`enumspecifier` holds three `enumerator`s, the first holding the literal `1`. The printer writes the body's items a
line each, as mbse-programs does; `Terms.walk` reads the tree, parents first:

````typescript
const enumerator = (name: string, value?: unknown) => (t: any) => t.enumerator(
  (r: any) => value === undefined ? r.name((i: any) => i.spelling(name)) : r.name((i: any) => i.spelling(name)).value(value));
const one = (t: any) => t.literal((r: any) => r.value((v: any) => v.as_scalar(1n)));
const [structure, color] = store.transaction(() => [
  S.C.Types.Expression((e: any) => e.root((u: any) => u.typeid((i: any) => i.specifiers([
    (s: any) => s.classspecifier((c: any) => c.key("struct").name((n: any) => n.identifier((i: any) => i.spelling("S"))).body(true))]))
  ).abi([amd64])),
  S.C.Types.Expression((e: any) => e.root((u: any) => u.typeid((i: any) => i.specifiers([
    (s: any) => s.enumspecifier((n: any) => n.key("enum").name((m: any) => m.identifier((i: any) => i.spelling("Color"))).body(true)
      .enumerators([enumerator("RED", one), enumerator("GREEN"), enumerator("BLUE")]))]))).abi([amd64])),
]);
console.log(CTypes.Text.ToText(structure));
console.log(CTypes.Text.ToText(color));
for (const t of Terms.walk(color)) {
  console.log(" ", Terms.kindOf(t)[0].padEnd(10), Terms.dialectOf(t), ...(Terms.nameOf(t) ? [Terms.nameOf(t)] : []));
}
assert.equal(CTypes.Text.ToText(structure), "struct S {}");
assert.equal(CTypes.Text.ToText(color), "enum Color {\n    RED = 1,\n    GREEN,\n    BLUE\n}");
````

An enumerator is a term of its own, so its holder may hold one alone, and it prints as it stands in a body. A body may
hold more than enumerators: the preprocessor's directives and comments stand among them, which
[8.08 · C Preprocessor](08_08_C_Preprocessor.ipynb) builds, so `C Types`' `Term` lists `C Preprocessor`'s directives
besides its own kinds (08.3 of tutorial 8):

````typescript
const alone = store.transaction(() => S.C.Types.Expression((e: any) => e.root(enumerator("RED", one)).abi([amd64])));
console.log(CTypes.Text.ToText(alone));
console.log(store.registered(DIALECT.union).branches.map((b: any) => b.name).join(", "));
assert.equal(CTypes.Text.ToText(alone), "RED = 1");
````

## 08.07.4 · Checked, and kept as written

`DIALECT.validate` checks a type name as C's grammar has it, and reports what is wrong by path, without evaluating: an
enumerator's name is an identifier, so `1A` is refused; a body's enumerators need the body (`enum E` declares none).
The terms a type name holds are C's, validated as C validates them, so an array's size or an enumerator's value that
reads `x` needs `x` from its scope (tutorial 4), as `Symbolics.free` says. `C Types` binds no name of its own: an
enumerator's name is declared, not bound within the term:

````typescript
const checked: Record<string, any> = store.transaction(() => ({
  "1A": S.C.Types.Expression((e: any) => e.root(enumerator("1A")).abi([amd64])),
  "enum E": S.C.Types.Expression((e: any) => e.root((u: any) => u.typeid((i: any) => i.specifiers([
    (s: any) => s.enumspecifier((n: any) => n.key("enum").name((m: any) => m.identifier((i: any) => i.spelling("E")))
      .enumerators([enumerator("A")]))]))).abi([amd64])),
  "int[x + 1]": S.C.Types.Expression((e: any) => e.root((u: any) => u.typeid((i: any) => i.specifiers([keyword("int")])
    .declarator((d: any) => d.arraydeclarator((a: any) => a.size((t: any) => t.binaryexpression((b: any) => b
      .left(variable("x")).operator("+").right(one))))))).abi([amd64])),
}));
for (const [source, name] of Object.entries(checked)) {
  console.log(source.padEnd(11), "|", DIALECT.validate(name).join("; ") || "valid",
    "| needs:", [...Symbolics.free(name, DIALECT)].sort().join(", ") || "nothing");
}
assert.deepEqual(DIALECT.validate(checked["1A"]), ["'1A' is not an identifier"]);
assert.deepEqual(DIALECT.validate(checked["int[x + 1]"], ["x"]), []);
````

A type name read from text that does not parse is kept as written ([PRI-002: preserve, don't lose](https://github.com/pitaman71/mbse-basis/blob/main/docs/PRINCIPLES.md#PRI-002)):
`int int` is no type name, and a reader that meets it writes `C Types`' `error`, with the text it skipped, which prints
back as written and which validation reports, here where a cast holds it:

````typescript
const skipped = store.transaction(() => S.C.Expressions.Expression((e: any) => e.root((u: any) => u.castexpression((r: any) => r
  .type((t: any) => t.error((x: any) => x.text("int int"))).operand(variable("x")))).abi([amd64])));
console.log(C.Text.ToText(skipped), "|", C.Expressions.DIALECT.validate(skipped, ["x"]).join("; "));
assert.deepEqual(C.Expressions.DIALECT.validate(skipped, ["x"]), ["type: invalid syntax 'int int'"]);
````

## 08.07.5 · Read by C, in a cast

```text
(uint8_t)(x + 1)        (unsigned char)300        sizeof(int)
```

C reads `C Types`' terms where its grammar puts a type, through the dialects it reads (tutorial 8, 08.1): a cast's
`type`, a compound literal's, a generic association's; and `typeid` stands among C's own kinds, where an expression or a
type may, in `sizeof`. So C validates, infers and evaluates a cast with its type in it: `(uint8_t)(x + 1)` is `0` when
`x` is 255, since the ABI's `uint8_t` wraps; `(unsigned char)300` is 44, and C infers its domain from the type it names,
the ABI's `unsigned char`. What evaluates a type is the cast that holds it: C's evaluator refuses `sizeof`, and a type
name alone, each by the name of the dialect whose kind it is:

````typescript
const uint8 = (t: any) => t.typeid((i: any) => i.specifiers([(s: any) => s.namedtypespecifier(
  (n: any) => n.name((m: any) => m.identifier((d: any) => d.spelling("uint8_t"))))]));
const [wrapped, narrowed, sized] = store.transaction(() => [
  S.C.Expressions.Expression((e: any) => e.root((u: any) => u.castexpression((r: any) => r.type(uint8)
    .operand((o: any) => o.binaryexpression((b: any) => b.left(variable("x")).operator("+").right(one))))).abi([amd64])),
  S.C.Expressions.Expression((e: any) => e.root((u: any) => u.castexpression((r: any) => r
    .type((t: any) => t.typeid((i: any) => i.specifiers([keyword("unsigned"), keyword("char")])))
    .operand((o: any) => o.literal((l: any) => l.value((v: any) => v.as_scalar(300n)))))).abi([amd64])),
  S.C.Expressions.Expression((e: any) => e.root((u: any) => u.sizeofexpression(
    (r: any) => r.operand((t: any) => t.typeid((i: any) => i.specifiers([keyword("int")]))))).abi([amd64])),
]);
for (const expression of [wrapped, narrowed, sized]) {
  const type = Terms.form(Terms.term(expression)).arguments[0]; // each one's first subterm, its type
  console.log(C.Text.ToText(expression).padEnd(18), Terms.kindOf(Terms.term(expression))[0].padEnd(16),
    "its type:", Terms.kindOf(type)[0], "of", C.Expressions.DIALECT.owner(type).name(),
    "| x = 255:", refused((c) => C.Evaluators.OfAny(c, { x: 255n }), expression));
}
console.log("C infers", C.Text.ToText(narrowed), "as", C.Expressions.DIALECT.infer(narrowed).name());
console.log(refused((x) => C.Evaluators.OfAny(x), int_));
assert.deepEqual([wrapped, narrowed, sized].map((x) => C.Text.ToText(x)), ["(uint8_t)(x + 1)", "(unsigned char)300", "sizeof(int)"]);
assert.ok(C.Evaluators.OfAny(wrapped, { x: 255n }) === 0n && C.Evaluators.OfAny(narrowed) === 44n);
````

`int` is the ABI's (02.3), so in a holder that names no ABI neither the literal `300`, an `int`, nor a cast to `int`
has a domain, and inference refuses the first it meets, the cast's operand:

````typescript
const unnamed = store.transaction(() => S.C.Expressions.Expression((e: any) => e.root((u: any) => u.castexpression((r: any) => r
  .type((t: any) => t.typeid((i: any) => i.specifiers([keyword("int")])))
  .operand((o: any) => o.literal((l: any) => l.value((v: any) => v.as_scalar(300n))))))));
console.log(refused((x) => C.Expressions.DIALECT.infer(x), unnamed));
````

## 08.07.6 · Its editions

`C Types` covers C's three editions, as C does (02.3), each by its designation and its common name, and records the
kinds each has ([CAP-268: standards as tagged data](https://github.com/pitaman71/mbse-basis/blob/main/docs/capabilities/CAP-268.md)):
both of its kinds are in all three, so no kind of it is newer than C11:

````typescript
const types = store.Instances.Expressions.C.Types; // the dialect, by its global name
for (const [designation, edition] of Standards.standards(types)) {
  console.log(designation, "|", Standards.tags(edition).slice(1).join(", "), "|",
    Standards.kinds(edition).map((kind: any) => kind.name).join(", "));
}
assert.ok(Standards.find(types, "C23") === Standards.standards(types).get("ISO/IEC 9899:2024"));
````

## What you learned

- **A type name is a `C Types` term**, `typeid`: its specifiers and its declarator are helpers, nested as C's
  declarators nest, and the terms they hold, an array's size, an enumerator's value, are C's.
- **An enumeration's enumerators are terms of their own**, `enumerator`, among which the preprocessor's directives may
  stand ([8.08 · C Preprocessor](08_08_C_Preprocessor.ipynb)).
- **Its holder names an ABI**, as C's does, since what `int` is depends on it.
- **It is checked and kept as written**: validation reports what C's grammar refuses, and text that does not parse is
  kept as `error`, printed back as written.
- **C reads it where a type stands**: a cast converts by it and infers its domain from it, while a type name itself is
  never evaluated.

C++'s type-ids are a dialect of their own, [8.09 · Cpp Types](08_09_Cpp_Types.ipynb); next comes the preprocessor that C
and C++ share.
